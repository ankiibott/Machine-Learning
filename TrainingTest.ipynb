```python
import pandas as pd

data = {
    'Mileage': [
        69000, 35000, 57000, 22500, 46000, 59000, 52000, 72000, 91000, 67000,
        83000, 79000, 59000, 58780, 82450, 25400, 28000, 69000, 87600, 52000
    ],
    'Age(yrs)': [
        6, 3, 5, 2, 4, 5, 5, 6, 8, 6,
        7, 7, 5, 4, 7, 3, 2, 5, 8, 5
    ],
    'Sell Price($)': [
        18000, 34000, 26100, 40000, 31500, 26750, 32000, 19300, 12000, 22000,
        18700, 19500, 26000, 27500, 19400, 35000, 35500, 19700, 12800, 28200
    ]
}
df = pd.DataFrame(data)
df.head()
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
#chart btw Independent and Dependent variable !!
plt.scatter(df['Mileage'],df['Sell Price($)'],marker='+')
```

```python
#chart btw Independent and Dependent variable !!
plt.scatter(df['Age(yrs)'],df['Sell Price($)'],color='g')
```

```python
X=df[['Mileage','Age(yrs)']]
y=df['Sell Price($)']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2) #20% testing
```

```python
X_train
```

```python
#now the same process 
from sklearn.linear_model import LinearRegression
clf = LinearRegression()
```

```python
clf.fit(X_train,y_train)
```

```python
clf.predict(X_test)
```

```python
clf.score(X_test,y_test) #94% accuracy
```
