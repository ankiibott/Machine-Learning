```python
import pandas as pd
df = pd.read_csv('C:\\Users\\ARPIT DUBEY\\salaries.csv')
df.head()
```

```python
df.rename(columns={
    'salary_more_then_100k': 'more_then_100k'
},inplace=True)
df.columns
```

```python
#indepedent variabble 
inputs= df.drop(['more_then_100k'],axis ='columns')
#dependent variable
target = df['more_then_100k']
```

```python
inputs.head()
```

```python
target.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
```

```python
inputs['company_n'] = le.fit_transform(inputs['company'])
inputs['job_n'] = le.fit_transform(inputs['job'])
inputs['degree_n'] = le.fit_transform(inputs['degree'])
```

```python
inputs_n = inputs.drop(['company','job','degree'],axis='columns')
```

```python
inputs_n
```

```python
from sklearn import tree
```

```python
model = tree.DecisionTreeClassifier()
```

```python
model.fit(inputs_n,target)
```

```python
model.score(inputs_n,target)
```

```python
model.predict([[2,2,0]])
```

```python
import pandas as pd 
d = pd.read_csv('C:\\Users\\ARPIT DUBEY\\titanic.csv')
d.head()
```

```python
df = d.drop(['PassengerId','Name','Ticket','SibSp','Parch','Cabin','Embarked'],axis='columns')
df.head()
```

```python
#independent variable
inputs = df.drop('Survived',axis='columns')
#dependent variable
target = df['Survived']
```

```python
target
```

```python
type(inputs['Sex']) # as this is a string & the model cant take string as an input , we need a numeric only input !!
```

```python
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

```python
inputs.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(inputs,target)
```

```python
from sklearn import tree
```

```python
model = tree.DecisionTreeClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict([[1,1,35,53.1]])
```
