```python
from sklearn.datasets import load_digits
import pandas as pd

dataset = load_digits()
dataset.keys()
```

```python
# as it has 64 columns
dataset.data.shape
```

```python
dataset.data[0]
```

```python
#converting into 2d array
dataset.data[0].reshape(8,8)
```

```python
from matplotlib import pyplot as plt
%matplotlib inline
plt.gray()
plt.matshow(dataset.data[0].reshape(8,8))
```

```python
df = pd.DataFrame(dataset.data, columns=dataset.feature_names)
df.head()
```

```python
dataset.target
```

```python
X = df
y = dataset.target
```

```python
#If X looked like this:
#[[10, 1000],
# [12, 1100],
 #[14, 1200]]
#After standardization, X_scaled might look like:
#[[-1.22, -1.22],
# [ 0.00,  0.00],
#[ 1.22,  1.22]]    

#this is what standardscaler is useful for , as it is better for the model to have same input!

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
```

```python
#NOW USING PCA TO REDUCE DIMENSIONS!!


X.shape
```

```python
#through PCA its clrly visible that the columns reduces to 29 
#0.95 means i need 95% useful features!!
from sklearn.decomposition import PCA

pca = PCA(0.95)
X_pca = pca.fit_transform(X)
X_pca.shape
```

```python
#this helps u identifying ki kaunsa column jyda info.(variance) dia h !
pca.explained_variance_ratio_
```

```python
pca.n_components_
```

```python
from sklearn.model_selection import train_test_split
X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)
```

```python
#as i said , through pca it is possible that the accuracy might dec, but ur computation will inc!!
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_pca, y_train)
model.score(X_test_pca, y_test)
```

```python
#Lets now select only two components !!


```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
X_pca.shape
```

```python
X_pca
```

```python
#You can see that both combined retains 0.14+0.13=0.27 or 27% of important feature information
pca.explained_variance_ratio_
```

```python
#the accuracy reduces more !!
#so  you have to choose by urself ki kitne components wud be enough to hv the best accuraxxy as well!

X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=30)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_pca, y_train)
model.score(X_test_pca, y_test)
```
