```python
%matplotlib inline
import matplotlib.pyplot as plt

from sklearn.datasets import load_digits
```

```python
digits = load_digits()
```

```python
dir(digits)
```

```python
digits.data[0]
```

```python
plt.gray()
for i in range(5):
    plt.matshow(digits.images[i])
```

```python
digits.target[0:5]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(digits.data,digits.target,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict([digits.data[76]])
```

```python
plt.matshow(digits.images[76])
```

```python
#through model.score we got 97% accuracy but now we need to find what is the mistake in the remaining 3 %
#we can achieve this through confusion_matrix
```

```python
y_predicted = model.predict(X_test)
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test,y_predicted)
cm
```

```python
#to have a better data visualisatio of the confusion_matrix we use SEABORN
import seaborn as sn 
plt.figure(figsize=(10,7))
sn.heatmap(cm,annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

```python
# now through matrix 
# '8' is the no. which got predicted wrong two times & model predicted here as '1'
# this means wherever there is no zero ,it means that is an error or the wrong prediction !!
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
dir(iris)
```

```python
iris.feature_names
```

```python
iris.data[0]
```

```python
iris.target_names
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris.data,iris.target,test_size=0.2)
```

```python
#total =150
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test) 
```

```python
y_predicted = model.predict(X_test)
from sklearn.metrics import confusion_matrix 
cm = confusion_matrix(y_test,y_predicted)
cm
```

```python
import seaborn as sn
```

```python
plt.figure(figsize=(3,3))
sn.heatmap(cm,annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```
