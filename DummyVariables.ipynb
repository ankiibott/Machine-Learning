```python
import pandas as pd
data = {
    'town': [
        'monroe township', 'monroe township', 'monroe township', 'monroe township', 'monroe township',
        'west windsor', 'west windsor', 'west windsor', 'west windsor',
        'robinsville', 'robinsville', 'robinsville', 'robinsville'
    ],
    'area': [
        2600, 3000, 3200, 3600, 4000,
        2600, 2800, 3300, 3600,
        2600, 2900, 3100, 3600
    ],
    'price': [
        550000, 565000, 610000, 680000, 725000,
        585000, 615000, 650000, 710000,
        575000, 600000, 620000, 695000
    ]
}
df = pd.DataFrame(data)
df
```

```python


#1st METHOD 
#through pandas get_dummies

```

```python
dummies = pd.get_dummies(df.town,dtype=int)
dummies
```

```python
merged = pd.concat([df,dummies],axis='columns')
merged
```

```python
#the reason of dropping 'town' is cuz , as we already have dummy variables columns so we can do the calc throught that 
#no need of extra column
#Reason of dropping one of this dummy variable column , i need to google this .But it has said that if you have 5 dummy columns then drop 
#one column ,,because ofmulticollinearity 
#DUMMY VARIABLE TRAP se bachne ke liye we drop smth like that !!

final = merged.drop(['town','west windsor'],axis='columns')
final
```

```python
#X is having independent variable
X= final.drop('price',axis='columns')
X
```

```python
y = final.price
y
```

```python
from sklearn import linear_model
model = linear_model.LinearRegression()
```

```python
model.fit(X,y)
```

```python
#town:-robbinsville
model.predict([[2800,0,1]])
```

```python
#twon:- wind 
model.predict([[2800,0,0]])
```

```python
#town:- monroe
model.predict([[2800,1,0]])
```

```python
#In order to ONE HOT ENCODING first you need to do labelencoding to that text independent column !

from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
#fit_transform takes the label column as an argument/input and return the labels 
#in this case , it is 0,2,1

dfle = df
le.fit_transform(dfle.town)
```

```python
dfle.town = le.fit_transform(dfle.town)
dfle
```

```python
#now the same process take X and y 
#the reason of taking X as values because we dont need X to return as a dataframe , instead we need it as a 2D array
X= dfle[['town','area']].values
X
```

```python
y = dfle.price
y
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer 

#creating an object so that baar baar itna bda naam na likhna pde

ohe = OneHotEncoder()


# apply OneHotEncoder to column index 0
column_transformer = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(), [0])
    ],
    remainder='passthrough'  # leave other columns as is
)

# Fit and transform your data
X = column_transformer.fit_transform(X)
```

```python
#the reason of using columnTransformer is because if in the nxt step if i put X as an input ohe.fit_transform()
#then it will consider all the columns as label which is eventually wrong 
# through this , we make sure ki joh label column h sirf whi consider ho which is in this case is 'town' the 0th column
#remainder ='passthrough' means the rest of the column shud not get affected through this columnTransformer
```

```python
X = X[:,1:] #takes all rows and drops oth column and considering columns starting from 1
X
```

```python
#Now the same step
model.fit(X,y)
```
