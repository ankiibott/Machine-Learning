```python
import pandas as pd 
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
dir(iris)
```

```python
iris.target_names
```

```python
iris.feature_names
```

```python
df = pd.DataFrame(iris.data,columns=iris.feature_names)
df.head()
```

```python
df['Target']= iris.target
df.head() #setosa
```

```python
df[df.Target==2].head() #virginca
```

```python
#adding a new column as a flower_name
df['flower_name'] = df.Target.apply(lambda x:iris.target_names[x])
df.head()
```

```python
from matplotlib import pyplot as plt
```

```python
%matplotlib inline
```

```python
#creating a separate dataframe for each flower
df0 = df[df.Target==0]
df1 = df[df.Target==1]
df2 = df[df.Target==2]
```

```python
plt.xlabel('sepal length(cm)')
plt.ylabel('sepal width(cm)')
plt.scatter(df0['sepal length (cm)'],df0['sepal width (cm)'],marker = '+',color='g')
plt.scatter(df1['sepal length (cm)'],df1['sepal width (cm)'],marker = '.',color='y')
```

```python
plt.xlabel('petal length(cm)')
plt.ylabel('petal width(cm)')
plt.scatter(df0['petal length (cm)'],df0['petal width (cm)'],marker = '+',color='g')
plt.scatter(df1['petal length (cm)'],df1['petal width (cm)'],marker = '.',color='y')
```

```python
x = df.drop(['Target','flower_name'],axis = 'columns')
```

```python
y = df['Target']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn.svm import SVC 
```

```python
#SVC :- support vector classifier the model 
model = SVC(C=10,kernel='linear')
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```
