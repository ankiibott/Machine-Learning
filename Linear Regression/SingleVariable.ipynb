```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
from sklearn import linear_model #scikit
```

```python
data = {
    'area': [2600, 3000, 3200, 3800, 4000],
    'price': [550000, 565000, 610000, 680000, 725000]
}
df = pd.DataFrame(data)
df
```

```python
%matplotlib inline
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
plt.title('Home Price')
plt.scatter(df.area,df.price,marker='+',color='g')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]])
```

```python
# as y=mx+b where x =3300 and the y is the output which is eventually we got as a price
#lets findyou what is the slop 
reg.coef_
```

```python
reg.intercept_
```

```python
%matplotlib inline
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
plt.title('Home Price')
plt.scatter(df.area,df.price,marker='+',color='g')
plt.plot(df.area,reg.predict(df[['area']]),color='y')
```
